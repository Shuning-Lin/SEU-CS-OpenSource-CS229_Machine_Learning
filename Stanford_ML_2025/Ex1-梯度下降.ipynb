# 简单的手搓版梯度下降

```python
import numpy as np
A=np.array([[1,0],[0,2]])
theta_0=np.array([[-1],[0.5]])  #我个人比较喜欢写成这种写法，虽然似乎并不是很正规

def J(theta):
    return theta.T @ A @ theta

def update_theta(theta,lr):
    gradient=(A+A.T) @ theta
    return theta-lr*gradient

def gradient_descend(J,update_theta,theta_0,lr,epsilon=1e-50):
    theta_t=theta_0.copy()  #注意浅拷贝的问题
    while True:
        J_t1=J(theta_t)
        theta_t=update_theta(theta_t,lr)
        J_t2=J(theta_t)
        if abs(J_t2-J_t1)<epsilon or J_t2>1e20:
            break
    return theta_t

if __name__ == "__main__":
    print("请输入你预期的学习率:")
    lr=float(input())
    theta=gradient_descend(J,update_theta,theta_0,lr)
    assert all(abs(theta_i)<1e-24 for theta_i in theta), f"elements of theta {theta} is to far from the optimal value"
    print("最终优化后的结果为:",J(theta))
    print("最终优化后的参数为:")
    print(theta)
```

# 其实也可以直接用SciPy求解

```python
from scipy.optimize import minimize

def grad(theta):
    return 2*(A @ theta)

#注意.ravel()函数的应用以及“BFGS”拟牛顿法 
result=minimize(fun=J,x0=theta_0.ravel(),jac=grad,method="BFGS",options={"gtol":1e-8,"maxiter":1000})
print("success",result.success)
print("theta",result.x)
print("loss",result.fun)
print("iterations",result.nit)
print("message",result.message)
```
