```python
import numpy as np
import matplotlib.pyplot as plt
from Ex3_reuse_module import read_csv
from Ex3_reuse_module import train
from Ex3_reuse_module import f


def main(train_path):
    train_inputs,train_labels=read_csv(train_path)
    degree=[3,5,10,20]
    colors=['blue','red','green','cyan']
    color_position=0
    fig=plt.figure()
    plt.scatter(train_inputs,train_labels,c='b',marker='o',label='train_data')
    interval=np.linspace(-6,6,500)
    for i in degree:
        X=[]
        for item in train_inputs:
            x=[]
            for j in range(i+1):
                x.append(item**j)
            X.append(x)
        theta=train(np.array(X),train_labels)
        plt.plot(interval,f(interval,theta,i),linestyle='-',color=colors[color_position],label=f"degree={i}")
        color_position+=1
        plt.xlabel('x')
        plt.ylabel('y')
        plt.legend()
    plt.savefig('Ex3_2-plot',dpi=300)
    plt.show()        
if __name__=="__main__":
    main('train.csv')
    
```
